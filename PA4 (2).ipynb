```python
import matplotlib.pyplot as plt
import pandas as pd
```

## PART A: VISCOMM DATAFRAME

```python
df = pd.read_excel("board2.xlsx")
df["Average"] = df[["Math", "GEAS", "Electronics", "Communication"]].mean(
    axis=1)
```

```python
vis_comm_filter = (df["Hometown"] == "Visayas") & (
    df["Track"] == "Communication")
```

```python
VisComm = df[vis_comm_filter][
    ["Name", "Gender", "Math", "Electronics", "Average"]
]
```

```python
print("=== PART A: VisComm ===")
print("Row count:", len(VisComm))
VisComm
```

## PART B: VISFEMALE DATAFRAME

```python
vis_female_filter = (df["Hometown"] == "Visayas") & (df["Gender"] == "Female")
VisFemale = df[vis_female_filter][
    ["Name", "Track", "GEAS", "Electronics", "Average"]
]
```

```python
print("=== PART B: VisFemale ===")
VisFemale
```

## PART C: CATEGORY MEANS & PLOTTING

```python
mean_track = df.groupby("Track")["Average"].mean()
mean_gender = df.groupby("Gender")["Average"].mean()
mean_hometown = df.groupby("Hometown")["Average"].mean()
```

```python
print("=== SUMMARY TABLES ===")
print("Mean by Track:", mean_track)
print("Mean by Gender:", mean_gender)
print("Mean by Hometown:", mean_hometown)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4), sharey=True)

mean_track.plot(
    kind="bar",
    ax=axes[0],
    title="Mean Average by Track",
    ylabel="Average Score",
    rot=0,
)

mean_gender.plot(
    kind="bar",
    ax=axes[1],
    title="Mean Average by Gender",
    ylabel="Average Score",
    rot=0,
)

mean_hometown.plot(
    kind="bar",
    ax=axes[2],
    title="Mean Average by Hometown",
    ylabel="Average Score",
    rot=0,
)

plt.tight_layout()
plt.show()
```

```python
plt.tight_layout()
plt.show()
```

```python
print("=== HIGHEST CATEGORY MEANS ===")
print(f"Track: {mean_track.idxmax()} ({mean_track.max():.2f})")
print(f"Gender: {mean_gender.idxmax()} ({mean_gender.max():.2f})")
print(f"Hometown: {mean_hometown.idxmax()} ({mean_hometown.max():.2f})")
```
